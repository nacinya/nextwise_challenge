```python
actions= {
     "Schneider Electric": "SU.PA",
    "Prysmian": "PRY.MI",
    "Legrand": "LR.PA",
    "ASML": "ASML.AS",
    "Adyen": "ADYEN.AS",
    "ASM International": "ASM.AS",
    "Nexans": "NEX.PA",
    "Thales": "HO.PA",
    "Dassault Aviation": "AM.PA",
    "Leonardo": "LDO.MI",
    "Technip Energies": "TE.PA",
    "Eramet": "ERA.PA",
    "SPIE": "SPIE.PA"
}

print ("Nombres d'actions:", len(actions))
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
tickers = list(actions.values())

data= yf.download(
    tickers,
    period= "3y",
    auto_adjust= True,
    progress= False
)

prices= data["Close"]
print(prices.shape)
display(prices.head())
```

```python
print(prices.isna().sum())
```

```python
returns = prices.pct_change(fill_method= None)
display(returns.head())
```

```python
volatility= returns.std() * np.sqrt(252) * 100
drawdown= (prices / prices.cummax()-1).min() *100
```

```python
risk_table= pd.DataFrame({
    "Volatilité (%)": volatility,
    "Max Drawdown (%)": drawdown
})

risk_table= risk_table.sort_values("Volatilité (%)")

display(risk_table.round(2))
```

```python
#calcule de la corrélation entre chaque paire d'actions
correlation= returns.corr()
display(correlation.round(2))
```

```python
# 1. SPIE et Legrand présentent une volatilité historique relativement faible parmi vos 13 actions.
# 2. Eramet, Adyen et ASM International présentent une volatilité plus élevée et des drawdowns historiques importants.
# 3. Plusieurs entreprises appartiennent à des groupes fortement corrélés, donc il faudra limiter la concentration dans ces groupes.
```

```python
# les actions historiquement moins volatiles reçoivent un poids plus important
budget= 0.65

inverse_vol= 1/volatility

weights = inverse_vol /inverse_vol.sum() * budget

allocation = pd.DataFrame({
    "Weight (%)" : weights *100,
    "Amount (€)": weights * 200000
})

display(allocation.sort_values("Weight (%)", ascending=False).round(2))
```

```python
clean_returns= returns.dropna()
cov_matrix= clean_returns.cov()* 252

display(cov_matrix.round(4))
```

```python
w = weights.reindex(cov_matrix.columns).to_numpy()
portfolio_vol = np.sqrt(w @ cov_matrix.to_numpy() @ w)*100

print(f"Portfolio volatility: {portfolio_vol:.2f}%")
```

```python
cov = cov_matrix.to_numpy()

high_weight_stocks= weights[weights > 0.05].index

high_mask = cov_matrix.columns.isin(high_weight_stocks)

def portfolio_variance(w):
    return w @ cov @ w

bounds = [
    (0.01, 0.10) if ticker in high_weight_stocks else (0.01, 0.05)
    for ticker in cov_matrix.columns
]
```

```python
constraints= [
    {
        "type": "eq",
        "fun": lambda w: np.sum(w) - 0.65
    },
    {
        "type": "ineq",
        "fun": lambda w: 0.40 - np.sum(w[high_mask])

    }
]
```

```python
from scipy.optimize import minimize
result = minimize(
    portfolio_variance,
    x0=w,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints,
    options={"ftol": 1e-12, "maxiter": 500}
)

print(result.success)
print(result.message)
```

```python
# Extract optimised portfolio weights
optimal_weights = pd.Series(result.x, index=cov_matrix.columns)

# Calculate optimised portfolio volatility
optimal_vol = np.sqrt(portfolio_variance(result.x)) * 100

print(f"Previous volatility: {portfolio_vol:.2f}%")
print(f"Optimised volatility: {optimal_vol:.2f}%")

# Check portfolio constraints
print(f"Total allocation: {optimal_weights.sum()*100:.2f}%")
print(f"Maximum weight: {optimal_weights.max()*100:.2f}%")
print(f"Weight above 5%: {optimal_weights[optimal_weights > 0.050001].sum()*100:.2f}%")

# Display optimised allocations
display((optimal_weights * 100).sort_values(ascending=False).round(2))
```

```python
previous_returns = clean_returns @ weights.reindex(clean_returns.columns)
optimised_returns = clean_returns @ optimal_weights.reindex(clean_returns.columns)

previous_sharpe = (
    previous_returns.mean() * 252
    / (previous_returns.std() * np.sqrt(252))
)

optimised_sharpe = (
    optimised_returns.mean() * 252
    / (optimised_returns.std() * np.sqrt(252))
)

print(f"Previous Sharpe: {previous_sharpe:.2f}")
print(f"Optimised Sharpe: {optimised_sharpe:.2f}")
```

```python
mean_returns = clean_returns.mean() * 252

mu = mean_returns.reindex(cov_matrix.columns).to_numpy()
def negative_sharpe(w):
    portfolio_return = w @ mu
    portfolio_risk = np.sqrt(w @ cov @ w)

    return -portfolio_return / portfolio_risk

sharpe_result = minimize(
    negative_sharpe,
    x0=optimal_weights.reindex(cov_matrix.columns).to_numpy(),
    method="SLSQP",
    bounds=bounds,
    constraints=constraints,
    options={"ftol": 1e-12, "maxiter": 500}
)

print(sharpe_result.success)
print(f"Maximum historical Sharpe: {-sharpe_result.fun:.2f}")
```

```python
sharpe_weights = pd.Series(
    sharpe_result.x,
    index=cov_matrix.columns
)

# Calculate portfolio volatility
sharpe_vol = np.sqrt(
    sharpe_weights.to_numpy() @ cov @ sharpe_weights.to_numpy()
) * 100

print(f"Portfolio volatility: {sharpe_vol:.2f}%")
print(f"Total allocation: {sharpe_weights.sum()*100:.2f}%")
print(f"Maximum weight: {sharpe_weights.max()*100:.2f}%")
print(f"Weight above 5%: {sharpe_weights[sharpe_weights > 0.050001].sum()*100:.2f}%")

display((sharpe_weights * 100).sort_values(ascending=False).round(2))
```
